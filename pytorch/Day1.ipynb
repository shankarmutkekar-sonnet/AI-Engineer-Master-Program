```python
import torch

x = torch.tensor([1, 2, 3])          # from Python list
x = torch.zeros(3, 4)                # 3x4 tensor of zeros
x = torch.ones(3, 4)
x = torch.randn(3, 4)                # standard normal random
x = torch.arange(10)                 # like range()
```

```python
x = torch.tensor([1, 2, 3], dtype=torch.float32)   # default float type
x = torch.tensor([1, 2, 3], dtype=torch.float16)   # half precision
x = torch.tensor([1, 2, 3], dtype=torch.bfloat16)  # brain float — common in training
x = torch.tensor([1, 2, 3], dtype=torch.int64)     # default int type (a.k.a. torch.long)

x.dtype       # inspect it
x.float()     # cast to float32
x.to(torch.bfloat16)  # cast to a specific dtype
```

```python
x = torch.tensor([1, 2, 3])
x.device                    # device(type='cpu')

x = x.to('cuda')            # moves tensor to GPU (triggers a host->device copy)
x = x.cuda()                # equivalent shorthand
x = x.cpu()                 # moves back to CPU (device->host copy)

x = torch.tensor([1,2,3], device='cuda')  # create directly on GPU, no copy needed
```
